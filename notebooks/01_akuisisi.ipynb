```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive?latitude=-8.65&longitude=115.21&start_date=2024-01-01&end_date=2024-12-31&daily=temperature_2m_max,temperature_2m_min,precipitation_sum&timezone=Asia%2FMakassar"
res = requests.get(url)
data = res.json()
df = pd.DataFrame(data['daily'])
df.to_csv('../data/raw/cuaca_singaraja.csv', index=False)
df.head()
```
