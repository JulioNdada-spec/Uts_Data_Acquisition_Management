```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive?latitude=-8.65&longitude=115.22&start_date=2024-01-01&end_date=2024-12-31&daily=temperature_2m_max,precipitation_sum&timezone=Asia%2FMakassar"
response = requests.get(url)
data = response.json()
print("Status:", response.status_code)
print("Jumlah hari:", len(data['daily']['time']))
```
